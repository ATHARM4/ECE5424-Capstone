## 1. Imports

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

cwd = Path.cwd().resolve()
if (cwd / "reports").exists():
    root = cwd
else:
    root = cwd.parent
```

## 2. Baseline model comparison

These are the main baseline results across logistic regression, random forest, and XGBoost.

```python
baseline = pd.read_csv(root / "reports" / "baseline_results.csv")
baseline
```

```python
baseline.sort_values("roc_auc", ascending=False)
```

The combined feature set performed the best overall, and logistic regression had the highest ROC AUC in the current baseline comparison.

## 3. Feature importance

The interpretation step was run on the best baseline model.

```python
importance = pd.read_csv(root / "reports" / "feature_importances.csv")
importance.head(15)
```

```python
top = importance.head(10).sort_values("importance")
plt.figure(figsize=(8, 5))
plt.barh(top["feature"], top["importance"])
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 10 Features")
plt.tight_layout()
plt.show()
```

Age and several diet/body-size related variables show up near the top, which matches the baseline model outputs.

## 4. Exploratory intervention results

These are simulations below.

```python
recommendations = pd.read_csv(root / "reports" / "recommendation_simulation.csv")
recommendations.head(15)
```

```python
top_recs = recommendations.head(10).sort_values("mean_risk_reduction")
plt.figure(figsize=(8, 5))
plt.barh(top_recs["intervention"], top_recs["mean_risk_reduction"])
plt.xlabel("Mean predicted risk reduction")
plt.ylabel("Intervention")
plt.title("Top Intervention Results")
plt.tight_layout()
plt.show()
```

In the current simulation results, weight and sodium changes tend to produce the largest predicted reductions. I would still treat these outputs as exploratory because the model is trained on observational data.

## 5. Final note

Overall, the project shows that lifestyle variables carry some signal for hypertension prediction, especially when combined with demographic features. However, it can be concluded that this signal is definitely weaker than demographic by quite a bit.
